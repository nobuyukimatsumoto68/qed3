# One-point function $\langle F^2\rangle$ vs $g^2$ (per $L$)

$\langle F^2\rangle$ = flowed magnetic-energy density from the shape-basis $F^2$ driver: the $\ell=0$ component of the **triangle$^2$** operator (smallest loop = elementary $F_{12}^2$), consolidated over its icosahedral orbit and time-averaged. Per config it is the h5 dataset `F` at index `ishape*n_lm + ilm` with `ishape=0` (triangle), `ilm=0` ($\ell=0,m=0$), divided by `Nt`. Errors = binned jackknife (binsize 50, matching the correlator analysis).

Flow = spatial-only (per-timeslice), `FLOW_TMAX=2`. Data: L1 has the full $g^2\in\{1,2,4,8,12\}$ scan; L2/L4 only at $g^2=8$.

```python
import glob
import h5py
import numpy as np
import matplotlib.pyplot as plt

Nt = 128
BIN = 50            # jackknife bin size (trajectory units), matches the GEVP analysis
IDX = 0             # F dataset index: ishape=0 (triangle), ilm=0 (l=0,m=0)

# Nf series style: distinct colour AND marker (colour-blind safe)
STYLE = {2: ('#d62728', 'o'), 4: ('#1f77b4', 's'), 6: ('#2ca02c', '^')}
```

```python
# ---- load <F^2> per ensemble from the per-config h5 (slow: reads every config once) ----
GSTR = {1: '1.000000', 2: '2.000000', 4: '4.000000', 8: '8.000000', 12: '12.000000'}

ens = []
for Nf in (2, 4, 6):
    for g in (1, 2, 4, 8, 12):
        ens.append((Nf, g, 1))
for Nf in (2, 4, 6):
    ens.append((Nf, 8, 2))
    ens.append((Nf, 8, 4))

def jackknife(v, b):
    n = len(v)
    nb = n // b
    if nb < 2:
        return v.mean(), v.std() / np.sqrt(max(n, 1))
    x = v[:nb*b].reshape(nb, b).mean(1)
    m = x.mean()
    jkm = np.array([np.delete(x, i).mean() for i in range(nb)])
    err = np.sqrt((nb-1)/nb * np.sum((jkm - m)**2))
    return m, err

res = {}   # (Nf, gsq, L) -> (mean, err, ncfg)
for Nf, g, L in ens:
    dd = f'data_Nf{Nf}_gsq{GSTR[g]}at0.200000nu01.000000nt128L{L}'
    fs = sorted(glob.glob(f'{dd}/glue_f2_shapes.[0-9]*.h5'), key=lambda p: int(p.split('.')[-2]))
    v = []
    for f in fs:
        try:
            with h5py.File(f, 'r') as h:
                v.append(h['F'][IDX] / Nt)
        except Exception:
            pass
    if not v:
        continue
    v = np.array(v)
    m, e = jackknife(v, BIN)
    res[(Nf, g, L)] = (m, e, len(v))
    print(f'Nf{Nf} gsq{g:<2} L{L}: Ncfg={len(v):5d}  <F^2>={m:.6e} +/- {e:.2e}')
```

```python
# ---- L = 1 : <F^2> vs g^2 (full gsq scan), one series per Nf ----
L = 1
plt.figure(figsize=(7.0, 5.0))
ax = plt.gca()
for Nf in (2, 4, 6):
    c, mk = STYLE[Nf]
    xs, ys, es = [], [], []
    for g in (1, 2, 4, 8, 12):
        if (Nf, g, L) in res:
            m, e, _ = res[(Nf, g, L)]
            xs.append(g); ys.append(m); es.append(e)
    ax.errorbar(xs, ys, yerr=es, color=c, marker=mk, ms=8, capsize=3, lw=1.4, label=f'Nf={Nf}')
ax.set_xscale('log', base=2)
ax.set_xticks([1, 2, 4, 8, 12])
ax.set_xticklabels(['1', '2', '4', '8', '12'])
ax.set_xlabel(r'$g^2$')
ax.set_ylabel(r'$\langle F^2\rangle$  (triangle$^2$, $\ell=0$, flowed)')
ax.set_title(r'One-point $\langle F^2\rangle$ vs $g^2$,  $L=1$')
ax.grid(alpha=0.25)
ax.legend()
plt.tight_layout()
plt.show()
```
