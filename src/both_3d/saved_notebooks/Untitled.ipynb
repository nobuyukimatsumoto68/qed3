```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
curved4 = np.loadtxt( "ov_prop_temporal_L4_Nt144_T16.000000.dat" )
curved1 = np.loadtxt( "ov_prop_temporal_L1_Nt144_T16.000000.dat" )
flat1 = np.loadtxt( "flat_ov_prop_temporal_L1_Nt144_T16.000000.dat" )
```

```python
plt.plot( np.abs(curved4.T[1][::2] - curved1.T[1][::2]), label='c' )
plt.plot( np.abs(curved4.T[1][::2] - flat1.T[1][::2]), label='f' )
plt.yscale("log")
plt.legend()
```

```python
plt.plot( curved4.T[1][::2], label='c4' )
plt.plot( curved1.T[1][::2], label='c1' )
plt.plot( flat1.T[1][::2], label='f' )

plt.legend()
plt.yscale("log")
```

```python
Nt=144
at=16/144

# tmp = np.arccosh(l1_1/l1_1[int(Nt/2)])
# yy = (-tmp[1:]+tmp[:-1])/at
```

```python
def meff( l1_1 ):
    tmp = np.arccosh(l1_1/l1_1[int(Nt/2)-1])
    yy = (-tmp[1:]+tmp[:-1])/at
    return yy
```

```python
plt.plot( meff( curved4.T[1][::2][1:] ), label='4' )
plt.plot( meff( curved1.T[1][::2][1:] ), label='1' )
plt.plot( meff( flat1.T[1][::2][1:] ), label='1f' )

plt.ylim(0.9, 1.1)
plt.legend()
```
